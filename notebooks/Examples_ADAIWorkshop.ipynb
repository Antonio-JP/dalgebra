#### Loading our package `dalgebra`

#### Building the algebraic set-up 

```python
## Creating the field with constant g_3
R_g3 = PolynomialRing(QQ, "g_3")
## Adding the differential structure
DR_g3 = DifferentialRing(R_g3, [0])
DF_g3 = DR_g3.fraction_field()

## Creating the field of functions with the p-weierstrass
DF.<eta> = DElliptic(DF_g3, "eta_p^2 - eta^3 - g_3")
eta_p = DF.gen_p()
g_3, = DF.base().gens()

## Creating the ring of differential operators
DOP.<z> = DifferentialPolynomialRing(DF)
```

#### Creating the differential operator $L$

```python
## Creating the differential operator
L = 8*z[3] - 30*eta*z[1] - 15*eta_p*z[0]
L
```

#### Computing the centralizer of $L$ (we know its level)

```python
%%time
M = 4
Us = tuple(L.monic().coefficient_full(z[i]) for i in range(L.order(z)-1))

P,Gs,_ = GetCentralizer(Us, M, starting_level=M, ignore_bound=True)
```

```python
Gs
```

#### Computing the B.C. ideal of $L$

```python
%%time
I = BC_ideal(L.monic(), Gs[1:], z)
I
```

#### Analyzing the curve in SageMath

```python
A.<m_1,m_2,l> = AffineSpace(QQ,3)
local_dict = {'l':l, 'm_1':m_1, 'm_2':m_2, 'g_3':4}
els = [sage_eval(str(el).replace("lambda_", "l").replace("mu_1", "m_1").replace("mu_2", "m_2"), locals=local_dict) for el in I.gens()]
curve = Curve(els, A)
curve.genus()
```
