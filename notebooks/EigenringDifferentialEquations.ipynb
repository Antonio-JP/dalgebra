# **The Eigenring for an ODO**

## Introduction

## Main definition

In this document we explore the idea of Eigenring as defined introduced in (*add citations*) and the connections with the package `dalgebra` where this notebook is located.

Let $\mathbb{K}$ be a differential field with field of constants $C$. Let $L$ be a linear ordinary differential operator of the following form
$$L = \partial^n + u_2\partial^{n-2}+\ldots+u_{n-1}\partial + u_n \in \mathbb{K}[\partial].$$

Then $R \in \mathbb{K}[\partial]$ is said to be in the eigenring (let us say it's an eigenoperator) of $L$ if there is another operator $S \in \mathbb{K}[\partial]$ such that $LR = SL$. Equivalently, if $L$ divides from the right to $LR$.

The eigenring is the set $\mathcal{E}(L)$ of eigenoperators of order smaller than $n$.

##### **Why do we bound the order of the operators?**

Let $R$ be an eigenoperator of $L$ with $ord(R) \geq ord(L)$. Let us write $R = QL+\tilde{R}$ the Euclidean division of $R$ by $L$. Then we have:
$$LR = LQL + L\tilde{R} = SL.$$
Hence $\tilde{R}$ is also an eigenoperator with cofactor $S-LQ$.

On the other hand, let us assume that now $\tilde{R}$ is an eigenoperator. Then using the same formulation we have
$$LR = LQL + L\tilde{R} = LQL + \tilde{S}L = (LQ+\tilde{S})L.$$
Hence $R$ is also an eigenoperator.

This equivalency implies that we do not need to study all eigenoperators, but only those of bound order with the order of $n$, or, similarly speaking, we only need to consider operators in $\mathbb{K}[\partial] \pmod{L}$.

##### **Why the name eigenoperator?**

Let $y$ be a solution of the operator $L$ (i.e., $L(y) = 0$), and let $R \in \mathcal{E}(L)$. Then $R(y)$ is again a solution of $L$. 

*Proof:* since $LR = SL$, then we have that $L(R(y)) = S(L(y)) = S(0) = 0$.

Moreover, $R$ is a $C$-linear action over the solution space of solutions, i.e., $R(cy) = cR(y)$ and $R(y_1+y_2) = R(y_1)+R(y_2)$.

*Proof:* $R$ is a linear differential operator. Hence $R(y_1+y_2) = R(y_1)+R(y_2)$ by definition. On the other hand, $R(cy) = (Rc)(y) = cR(y)$.

Hence, $R$ have constant eigenvalues as a $C$-linear action over the solutions of $L$, and for each eigenvalue, the corresponding eigenfunction satisfies $R(f) = \lambda f$. The **eigenfucntions of $R$ are solutions of $L$.

*Proof:* $R(f) = \lambda f$, then 
$$L(f) = L(\frac{\lambda f}{\lambda}) = \frac{1}{\lambda}L(\lambda f) = \frac{1}{\lambda}L(R(f)) = \frac{1}{\lambda}S(L(f)),$$
rearranging this identity, we get
$$\lambda L(f) = S(L(f)),$$
or, eqiuvalently,
$$(\lambda L - SL)(f) = 0.$$

## Conditions for Eigenring

```python
from dalgebra import *
RO = DifferentialRing(QQ['c','d','e','x'], [0,0,0,1])
DOP.<u_2,u_3,v_0,v_1,v_2,z> = DifferentialPolynomialRing(RO.fraction_field())
c,d,e,x = DOP.base().gens()
L = z[3] + u_2*z[1]+u_3*z[0]
R = v_2*z[2]+v_1*z[1]+v_0*z[0]
```

```python
LR = L.dot(R, z)
```

```python
LR.coefficients(z), LR.monomials(z)
```

```python
LR1 = LR - v_2*L.dot(z[2],z)
LR1.coefficients(z), LR1.monomials(z)
```

```python
LR2 = LR1 - (v_1[0] + 3*v_2[1])*L.dot(z[1],z)
LR2.coefficients(z), LR2.monomials(z)
```

```python
LR3 = LR2 - (v_0[0] + 3*v_1[1] + 3*v_2[2])*L
LR3.coefficients(z), LR3.monomials(z)
```

```python
[el(v_2=1) for el in LR3.coefficients(z)]
```

```python
## If we have an order 0 operator, the conditions lead to a constant operator
[el(v_2=0,v_1=0) for el in LR3.coefficients(z)] 
```

```python
## For order 1 operator, we get v_1'' = -v_0' --> v_1' = -v_0 - c --> v_0 = -v_1' + c
## Then we get a system:
## v_1''' = -u_2*v_1'             --> v_1^(4) = -u_2'*v_1 -u_2*v_1''
## v_1^(4) = -3u_3v_1'-u_2*v_1''  --> -u_2'*v_1 -u_2*v_1'' = -3u_3v_1'-u_2*v_1''
v1 = v_1[0] 
v0 = -v_1[1] + c 
[el(v_2=0, v_1=v1, v_0=v0) for el in LR3.coefficients(z)] 
```
