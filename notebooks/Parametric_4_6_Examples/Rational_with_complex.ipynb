# A Parametric Rational Example for $n=4$ and $m=6$

When looking for non-trivial centralizers for operators of order $4$, we found a set of cases that can be parametrized using a parameter $t$.

$$L_1 = \partial^4 + \frac{t^2 - 15}{2x^2}\partial^2 + \frac{-t^2 + 8it + 15}{x^3}Dx + \frac{t^4 + 58t^2 - 192it - 135}{16x^4}$$
$$L_2 = \partial^4 + \frac{t^2 - 15}{2x^2}\partial^2 + \frac{-t^2 - 8it + 15}{x^3}Dx + \frac{t^4 + 58t^2 + 192it - 135}{16x^4}$$

In this notebook we are interested in the following, for each of the families:
* Compute the generic operator $B_6$ in the centralizer of order $6$.
* Compute the differential subresultant sequence between $L$ and $B_6$.

## Setting up the notebook

```python
import sys
sys.path.insert(0, "../..") # dalgebra is here

from dalgebra import *
from dalgebra.commutators import *

## Creating the gaussian rationals as the base field
QQi = NumberField(QQ["i"]("i^2+1"), "i") 
i = QQi.gen()

B.<x,t> = PolynomialRing(QQi)
C = PolynomialRing(QQi,t)
DB = DifferentialRing(B, [1,0])
DB.set_constant(DifferentialRing(C))
DF = DB.fraction_field()
DOP.<z> = DifferentialPolynomialRing(DF)
DOP

%display latex
```

## 1. The case of $L_1$

```python
L = z[4] + (t^2 - 15)/(2*x^2)*z[2] + (-t^2 + 8*i*t + 15)/(x^3)*z[1] + (t^4 + 58*t^2 - 192*i*t - 135)/(16*x^4)*z[0]
```

### 1.1. Computing the centralizer for the generic operator

```python
%%time
Us = tuple(L.coefficient_full(z[i]) for i in range(3))
_,Gs,_ = GetCentralizer(Us, 6, starting_level=6, ignore_bound=True)
B_6 = Gs[-1]
```

```python
%%time
BC_ideal(L, Gs[1:], z)
```

It seems this operator has a trivial centralizer up to level 12

### 1.2. Computing the subresultant sequence

```python
L_l, B_6_m = spectral_operators(L, B_6, names=["lambda_", "mu"])
SR = L_l.parent()
sz = SR.gen("z")
```

```python
L_l
```

```python
B_6_m
```

```python
%%time
## This computation breaks the kernell!!
SRS = SR.sylvester_subresultant_sequence(L_l, B_6_m, sz)
```

```python
for j in range(len(SRS)):
    show(j, LatexExpr(r"\longrightarrow"), SRS[j])
```

#### Analyzing the subsequence

##### The resultant

This case coincides with the BC ideal we computed before where $\mu^2 = \lambda^3$:

```python
SRS[0].coefficients()[0].numerator().wrapped.factor()
```

```python
I = [SRS[0].coefficients()[0].numerator().wrapped.factor()[0][0]]
```

##### The first subresultant

This case should yield an operator of order 1, but due to the structure of the problem it is a non constant order zero operator. We reduce it using the spectral curve (which does not depend on $t$):

```python
SR1_C = [el.numerator().wrapped.reduce(I) for el in SRS[1].coefficients()]
SR1_C
```

##### The second subresultant

```python
SR2_C = [el.numerator().wrapped.reduce(I) for el in SRS[2].coefficients()]
```

```python
P_SR2_C = SR2_C[0].parent()
PX = P_SR2_C.gens()[0]
I_tlm = ideal(sum([[P_SR2_C(coeff) for coeff in el.polynomial(PX).coefficients()] for el in SR2_C],[]))
```

```python
[el.polynomial(PX) for el in SR2_C]
```

```python
## We can alayze the ideal of conditions on t, lambda_ and mu to have a zero subresultant:
GB = I_tlm.groebner_basis()
GB
```

This means that we have (by the equation $\lambda^2)$ that $\lambda = 0$.

```python
[E for E in [el(lambda_=0) for el in GB] if E != 0]
```

So we have two cases: either $\mu = 0$ and $\lambda = 0$, or we have $t=0$ and $\lambda = 0$.

##### The third subresultant

```python
SR3_C = [el.numerator().wrapped.reduce(I) for el in SRS[3].coefficients()]
```

```python
P_SR3_C = SR3_C[0].parent()
PX = P_SR3_C.gens()[0]
I_tlm = ideal(sum([[P_SR3_C(coeff) for coeff in el.polynomial(PX).coefficients()] for el in SR3_C],[]))
```

```python
[el.polynomial(PX) for el in SR3_C]
```

```python
I_tlm.groebner_basis()
```
