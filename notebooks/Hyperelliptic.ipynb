## 1. Introduction

Hyperelliptic extensions are a type of differential extensions for differential fields where we introduce a new transcendental element $t$ whose derivative $t' = y$ is algebraic over the base field $\mathbb{K}(t)$, where $(\mathbb{K}, \partial)$ is a known differential field.

This is the essence of these extensions and from here we can build the structure in SageMath to work with them. This notebook is focused on the study of some properties for these extensions.

```python
R = DElliptic(DifferentialRing(QQ[x], (1,)).fraction_field(), "t_p^2 - t^3 - 1",  "t")
x = R.base().gen(0)
t = R.gen()
y = R.gen_p()
```

### 1.1 The algebraic structure

In general, let $p(y) \in \mathbb{K}(t)[y]$ be the minimal polynomial for $y=t'$. This polynomial is monic and irreducible. Let us write 
$$p(y) = \alpha_0(t) + \alpha_1(t) y + \ldots + y^n,$$
where $n$ is the degree of the extension.

Hence, the quotient $\mathbb{K}(t)[y]/(p(y))$ is a field extention of $\mathbb{K}(t)$ that can be seen as a $n$-dimensional vector space over $\mathbb{K}(t)$. All the elements can be associated with a vector $(f_0(t),\ldots,f_{n-1}(t)) \in \mathbb{K}(t)^n$ where the associated element is the combination of the classical basis $\{1, y, y^2, \ldots, y^{n-1}\}$.

In here, we have that the multiplication by $y$ is a $\mathbb{K}(t)$-linear map with the companion matrix as its representation:
$$M_y = C_{p(y)} = \begin{pmatrix}0 & 0 & \ldots & 0 & -\alpha_0(t)\\1 & 0 & \ldots & 0 & -\alpha_1(t)\\0 & 1 & \ldots & 0 & -\alpha_2(t)\\\vdots&\vdots&\ddots&\vdots&\vdots\\0 & 0 & \ldots & 1 & -\alpha_{n-1}(t)\end{pmatrix}.$$

Similarly, if we have an element $f(y) = f_0(t) + \ldots f_{n-1}(t)y^{n-1}$, the multiplication by $f(y)$ is a $\mathbb{K}(t)$-lienar map with matrix representation $M_{f(y)}= f(M_y)$.

### 1.2 The differential structure

Extending the differential structure is guided by the fact that $t' = y$ and that $y$ is algebraic over $\mathbb{K}(t)$ with minimal polynomial $p(y)$. Similar to algebraic extensions, where we can always extend uniquely the derivative to an algebraic element, we can do similar here:

* Let $p(y)$ be the minimal polynomial for $y$.
* Let $p_y(y)$ be the partial derivative of $p(y)$ w.r.t. $y$, i.e., the derivation where $\partial_y (\alpha) = 0$ for $\alpha\in \mathbb{K}(t)$ and $\partial_y(y) = 1$.
* Let $\beta p (y)$ be the coefficient derivative of $p(y)$, i.e., the derivation where $\beta(\alpha) = \partial(\alpha)$ for $\alpha \in \mathbb{K}(t)$, $\beta(y) = 0$.

Then, it is easy to check that:

* $\partial(f(y)) = \beta f(y) + y' f_y(y)$ for any $f(y) \in \mathbb{K}(t)[y]$.
* $p_y(y)$ has degree $n-1$ as a polynomial.
* $\gcd(p(y), p_y(y)) = 1$, i.e., there are polynomials $r(y),s(y)$ with $\deg(s) < n$ such that $$r(y)p(y) + s(y)p_y(y) = 1.$$
* Taking the quotient by $(p(y))$, we conclude that $s(y)p_y(y) \equiv 1\pmod{p(y)}$.
* Hence, $y' = -s(y)\beta p(y)$.

Let us denote by $\tilde{s}(y)$ be a polynomial of degree at most $n-1$ such that $-s(y)\beta p(y) \equiv \tilde{s}(y)\pmod{p(y)}$.

**Theorem** There is a unique differential extension of $(\mathbb{K}, \partial)$ such that $t$ is a transcendental element whose derivation is algebraic with minimal polynomial $p(y)$.

### 1.3 Computing $\tilde{s}(y)$

We have two possible ways (equivalent) to compute the polynomial $\tilde{s}(y)$ which represents the derivative of $y$ (i.e., the second derivative of $t$). The first method is the direct method (that comes naturally from the definition). Namely, we compute with the Extended Euclidean Algorithm the polynomial $s(y)$, then we compute $s(y)\beta p(y)$ and finally we reduce the result using $p(y) = 0$, i.e., $y^n = -\alpha_0(t) - \alpha_1(t)y -\ldots-\alpha_{n-1}(t)y^n-1$.

The second method is based on linear algebra over $\mathbb{K}(t)$. Let $\tilde{s}(y) = a_0(t) + \ldots + a_{n-1}(t)y^{n-1}$. Then we know that $y' = \tilde{s}(y)$.

If we now compute the derivative of the minimal polynomial, we get:
$$0 = \partial(p(y)) = \beta p(y) + \tilde{s}(y) p_y(y) = \kappa p(y) + y p_t(y) + \tilde{s}(y)p_y(y),$$
where $\kappa$ is a derivation such that $\kappa(\alpha) = \alpha'$ for $\alpha \in \mathbb{K}$, $\kappa(t) = 0$ and $\kappa(y) = 0$; and $p_t(y)$ is the derivative w.r.t. $t$, i.e., $\partial_t(\alpha) = 0$ for $\alpha \in \mathbb{K}$, $\partial_t(t) = 1$ and $\partial_t(y) = 0$.
