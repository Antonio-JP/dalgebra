### Example 3.4, 5.4 and 5.5 (Hyperbolic example)

Let us consider the differential operator of order 3
$$L = \partial_x^3 + \frac{6}{\cosh(x)^2}\partial_x.$$

```python
B.<cosh> = DElliptic(DifferentialRing(QQ), "cosh^2 - cosh_p^2 - 1")
sinh = cosh.derivative()
DOP.<z> = DifferentialPolynomialRing(B)
L = z[3] + 6/cosh^2 * z[1]
L
```

Now we can compute its centralizer. Using previous results, we know the level of this operator is 4:

```python
_,G,_ = GetCentralizer(tuple(L.coefficient_full(z[i]) for i in range(3-1)), 4, starting_level=4, ignore_bound=True)
```

For this operator we have three generators of the centralizer as a $C[L]$-module. Their orders are, respectively, congruent with 0,1 and 2 (modulo 3):

```python
G[0]
```

```python
G[1]
```

```python
G[2]
```

We can now compute the Burchnall-Chaundy ideal (its Gröbner basis) using the method `BC_ideal`:

```python
%%time
I = BC_ideal(L, G[1:], z)
for el in I.gens():
    show(el)
```

### Example 3.5 and 5.7 (Elliptic example)

Let us consider the differential operator of order 4
$$L = \partial_x^4 - 12\wp(x)\partial_x^2 + 1.$$

```python
C = DifferentialRing(QQ["g_2","g_3"]).fraction_field()
g_2,g_3 = C.gens()
B.<eta> = DElliptic(C, "eta_p^2 - 4*eta^3 - g_2*eta - g_3")
DOP.<z> = DifferentialPolynomialRing(B)
L = z[4] -12*eta* z[2] + z[0]
L
```

Now we can compute its centralizer. Using previous results, we know the level of this operator is 4:

```python
_,G,_ = GetCentralizer(tuple(L.coefficient_full(z[i]) for i in range(4-1)), 5, starting_level=5, ignore_bound=True)
```

For this operator we have three generators of the centralizer as a $C[L]$-module. Their orders are, respectively, congruent with 0,1 and 2 (modulo 3):

```python
G[0]
```

```python
G[1]
```

```python
G[2]
```

```python
G[3]
```

We can now compute the Burchnall-Chaundy ideal (its Gröbner basis) using the method `BC_ideal`:

```python
%%time
I = BC_ideal(L, G[1:], z)
BC_R = I.ring()
for el in I.gens():
    show(el)
```

### Example 5.6

We provide an example where the Burchnall-Chaundy ideal of an ODO is larger than the radical ideal of the differential resultants of Goodearl's basis. This computation (the radical ideal) takes big amounts of time, hence we take a specific example for an elliptic function and a rational function, so the computations runs smoothly.

##### Elliptic example (not enough with resultants)

We create the set up for the computations. In this case, we start from an operator $L$ of order 4 with coefficients in the differential field $\mathbb{Q}\langle \wp\rangle$, where $\wp$ is the $\wp$-Weierstrass function defined with the equation:
$$\wp'^2 = \wp^3 + 1.$$
Then, we have the operator 
$$L = \partial^4 - 3\wp \partial^2 - 6\wp' \partial - \frac{9}{2}\wp^2.$$

```python
B = DifferentialRing(QQ)
D.<eta> = DElliptic(B, "eta_p^2 - eta^3 - 1")
eta_p = eta.derivative()
Us = (-9/2*eta^2, -6*eta_p, -3*eta)
L, (G_0,G_1,G_2,G_3),_ = GetCentralizer(Us, 10, starting_level=5, ignore_bound=True)
z = L.parent().gen("z")
```

This operator has a non-trivial centralizer which $C[L]$-module structure is generated by the operators:

```python
G_0
```

```python
G_1
```

```python
G_2
```

```python
G_3
```

We now compute the Burchnall-Chaundy ideal for this operator. We do so with our method `BC_ideal`:

```python
I = BC_ideal(L, [G_1,G_2,G_3], L.parent().gen("z"))
BC_R = I.ring()
for el in I.gens():
    show(el)
```

We compute now the spectral curves associated with each pair of operators in Goodearl basis. We use `spectral_operators` to generate the spectral parameters and the new operators and then we use differential resultants to compute the corresponding polynomials generated by all these operators.

We check that these differential resultats are contained in the BC-ideal, as, by definition, are annihilated by the operators in Goodearl's basis:

```python
%%time
from dalgebra.commutators.spectral import *
spectral_ops = spectral_operators(G_1,G_2,G_3,L, names=["mu_1", "mu_2", "mu_3","lambda_"])
spectral_curves = {(i,j) : spectral_ops[i].sylvester_resultant(spectral_ops[j], "z") for i in range(len(spectral_ops)) for j in range(i+1,len(spectral_ops))}
all(I.reduce(BC_R(value)) == 0 for value in spectral_curves.values())
```

```python
for k,v in spectral_curves.items():
    print(k, latex(v))
```

```python
for k,v in spectral_curves.items():
    show(k, ":", v)
```

We compute know the radical ideal and a Gröbner basis of it, so we can check the ideal membership of all generators of the BC-ideal:

```python
I_res = ideal(BC_R, list(spectral_curves.values()))
I_rad = ideal(I_res.radical().groebner_basis())
```

We now reduce the BC-ideal using this Gröbner basis:

```python
[I_rad.reduce(p) for p in I.basis]
```

We see that these polynomials **do not** reduce to zero, hence they are not in the BC-ideal, showing that the radical ideal of the differential resultants is not enough to capture all the algebraic relations of the elements in Goodearl's basis.

##### (EXTRA) Rational example (enough with resultants)
