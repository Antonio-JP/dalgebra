We build the ring where spectral operators will exist:

```python
FBS = QQ["mu_1", "mu_2", "lambda_"]
IS = I.change_ring(FBS)
CD = FBS.quotient(IS)
DCD = DifferentialRing(CD).fraction_field()
mu_1, mu_2, l = DCD.gens()
RS.<eta> = DElliptic(DCD, "eta_p^2 - 4*eta^3 - 1")
eta_p = eta.derivative()
mu_1, mu_2, l = RS(mu_1), RS(mu_2), RS(l)
DOS.<z> = DifferentialPolynomialRing(RS)
eta, eta_p = DOS(eta), DOS(eta_p)
mu_1, mu_2, l = DOS(mu_1), DOS(mu_2), DOS(l)
```

```python
L = z[3] - 24*eta*z[1] - (12*eta_p + l)*z[0]
```

```python
G = (160*eta*eta_p - mu_1)*z[0] + (40*eta^2)*z[1] - (60*eta_p)*z[2] - (40*eta)*z[3] + z[5]
```
