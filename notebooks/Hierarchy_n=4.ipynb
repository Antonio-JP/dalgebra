```python
from dalgebra.pseries.laurent import LSChangeBound
LSChangeBound(5)

import re
reg_u = r"u_(\d+)_(\d+)"
reg_u = re.compile(reg_u)
def replacer_u(match):
    n,m = [int(el) for el in match.groups()]
    if m == 0:
        return f"sols[{n-2}]"
    elif m == 1:
        return f"sols[{n-2}].derivative()"
    else:
        return f"sols[{n-2}].derivative(times={m})"

reg_z = re.compile(r"z_(\d+)")
def replacer_z(match):
    n = int(match.groups()[0])
    return f"z[{n}]"
    
def evaluate_equation(equation, **kwds):
    locals_dict = {**kwds}
    command = reg_u.sub(replacer_u, str(equation))
    command = reg_z.sub(replacer_z, command)
    res = sage_eval(command, locals=locals_dict)
    return res
```

```python
reg = r"u(\d+)_(\d+)"
def get_coeff(n,m):
    if n < 2 or n > 4:
        raise ValueError("Incorrect value for index 'n'")
    elif m < 0:
        return C_e.zero()

    current_var = f"u{n}_{m}"

    if not current_var in coeffs:
        RHS = rhs[n-2].derivative(times=m-equs[n-2].order(u[n-2])) # rhs of the equation u_n^{(m)} = ...
        RHS_poly = RHS.to_sage().polynomial()
        var_names = [str(v) for v in RHS_poly.variables()]
        required = [match(reg, var).groups() for var in var_names]
        # we compute previous required elements
        for (i,j) in required:
            get_coeff(int(i),int(j))
        RHS_eval = C_e(RHS_poly(**{k:v for (k,v) in coeffs.items() if k in var_names}))
        coeffs[current_var] = RHS_eval/lhs[n-2]
    return coeffs[current_var]
```

```python
from functools import lru_cache
@lru_cache
def div_derivative(num,den,times:int):
    if times == 0:
        return (num, den)
    elif times == 1:
        return (num.derivative()*den - den.derivative()*num), den**2
    else:
        ders = {k : div_derivative(num,den,k) for k in range(times)}
        return num.derivative(times=times)*den**times - sum(ders[k][0]*den**(times-1-k)*den.derivative(times=times-k)*binomial(times,k) for k in range(times)), den**(times+1)

LS = LaurentSeries(DifferentialRing(QQ), 't')
t = LS.gen()
base = t+t^4

@lru_cache
def init_values(equation, gen, values, field = None):
    o = equation.order(gen)
    if equation.degree(gen[o]) != 1:
        raise ValueError(f"Necessary a linear equation in its highest degree")
    field = field if field is not None else equation.parent().constant_ring()
    LS = LaurentSeries(field, 't')
    t = LS.gen()
    base = sum(values[i]/factorial(i)*t**i for i in range(len(values)))
    
    lc = equation.coefficient_full(gen[o])
    den_ev = field(lc(**{gen.variable_name(): base}).coefficients()[0][0]) # this is in the constant field

    num = lc*gen[o] - equation
    computed = {i : values[i] for i in range(len(values))}
    def __init_values(k: int):
        global base
        if k < 0: return field.zero()
        elif k not in computed:
            for i in range(k): __init_values(i) # we ensure computation of previous values

            num_d,den = div_derivative(num, lc, k)
            num_d_ev = field(lc(**{gen.variable_name(): base}).coefficients()[0][0])
            den_evev = den_ev**(k+1)
            new_val = num_d_ev / den_evev
            base += t**(k) * new_val / factorial(k)
            computed[k] = new_val

        return computed[k]
    return __init_values


            
    
```

```python
H = dict()
for i in range(10):
    _,H[i] = almost_commuting_wilson(3,i)

P = H[0][0].parent()
C = P.constant_ring().to_sage()
u = P.gens()[:-1]
z = P.gens()[-1]
```

From equation 2, we can see that we can compute $u_2^{(i)}(0)$ for any $i \geq 4$ knowing the 8 values $u_2(0), u_2'(0), u_2''(0), u_2'''(0), u_3(0), u_3'(0), u_3''(0), u_3'''(0)$. Our goal is to find another equation that allows to compute $u_3^{(j)}(0)$ for $j \geq 4$ using the same 8 coefficients (or less).

```python
[[H[4][i].coefficient_full(ui[H[4][i].order(ui)]) for ui in u] for i in range(2)]
```

```python
(H[4][1].derivative()*2/3 - H[4][0]).orders()[:2]
```

This equation is preciesely what we were looking for. Here we can obtain higher derivtives of $u_3$ using the same 8 initial conditions as before.

```python
## equs[0] allows to compute u_2 and equs[1] allows to compute u_3
equs = [H[4][1], (H[4][1].derivative()*2/3 - H[4][0])]
```

```python
R_C = H[4][0].parent().add_constants(*[f"a_{i}_{j}" for i in range(2,4) for j in range(4)])
u_c = R_C.gens()[:-1]
C_e = R_C.constant_ring()
a = C_e.gens()
```

```python
hs = [(equ.coefficient_full(ui[equ.order(ui)]), equ.coefficient_full(ui[equ.order(ui)])*ui[equ.order(ui)] - equ) for (equ,ui) in zip(equs, u_c)]

lhs = [C_e(hs[i][0]) for i in range(len(hs))]
rhs = [hs[i][1] for i in range(len(hs))]
```

```python
base = {(i,j) : C_e(f"a_{i}_{j}") for i in range(2,4) for j in range(4)}
coeffs = {f"u{i}_{j}": C_e(base[(i,j)]) for i in range(2,4) for j in range(4)}
```

```python
reg = r"u(\d+)_(\d+)"
def get_coeff(n,m):
    if n < 2 or n > 4:
        raise ValueError("Incorrect value for index 'n'")
    elif m < 0:
        return C_e.zero()

    current_var = f"u{n}_{m}"

    if not current_var in coeffs:
        RHS = rhs[n-2].derivative(times=m-equs[n-2].order(u[n-2])) # rhs of the equation u_n^{(m)} = ...
        RHS_poly = RHS.to_sage().polynomial()
        var_names = [str(v) for v in RHS_poly.variables()]
        required = [match(reg, var).groups() for var in var_names]
        # we compute previous required elements
        for (i,j) in required:
            get_coeff(int(i),int(j))
        RHS_eval = C_e(RHS_poly(**{k:v for (k,v) in coeffs.items() if k in var_names}))
        coeffs[current_var] = RHS_eval/lhs[n-2]
    return coeffs[current_var]
```
