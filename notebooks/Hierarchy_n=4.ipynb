```python
hs = [(equ.coefficient_full(ui[equ.order(ui)]), equ.coefficient_full(ui[equ.order(ui)])*ui[equ.order(ui)] - equ) for (equ,ui) in zip(equs, u)]

lhs = [C(hs[i][0]) for i in range(len(hs))]
rhs = [hs[i][1] for i in range(len(hs))]
new_constants = [f"a_{i}_{j}" for i in range(2,5) for j in range(5)]
C_e = PolynomialRing(C, new_constants)
```

```python
C_e
```

```python
base = {(i,j) : C_e(f"a_{i}_{j}") for i in range(2,5) for j in range(5)}
coeffs = {f"u{i}_{j}": C_e(base[(i,j)]) for i in range(2,5) for j in range(5)}
```

```python
reg = r"u(\d+)_(\d+)"
def get_coeff(n,m):
    if n < 2 or n > 4:
        raise ValueError("Incorrect value for index 'n'")
    elif m < 0:
        return C_e.zero()

    current_var = f"u{n}_{m}"

    if not current_var in coeffs:
        RHS = rhs[n-2].derivative(times=m-equs[n-2].order(u[n-2])) # rhs of the equation u_n^{(m)} = ...
        RHS_poly = RHS.to_sage().polynomial()
        var_names = [str(v) for v in RHS_poly.variables()]
        required = [match(reg, var).groups() for var in var_names]
        # we compute previous required elements
        for (i,j) in required:
            get_coeff(int(i),int(j))
        RHS_eval = C_e(RHS_poly(**{k:v for (k,v) in coeffs.items() if k in var_names}))
        coeffs[current_var] = RHS_eval/lhs[n-2]
    return coeffs[current_var]
```

## Example with $n=2$

```python
H = dict()
for i in range(10):
    _,H[i] = almost_commuting_wilson(2,i)

P = H[0][0].parent()
C = P.constant_ring().to_sage()
u = P.gens()[:-1]
z = P.gens()[-1]
```

```python
H[3][0]
```

We need three initial values to define any solution: we create the three constants:

```python
R_C = H[3][0].parent().add_constants(*[f"a_{i}" for i in range(3)])
u_c = R_C.gens()[:-1]
C_e = R_C.constant_ring()
a = C_e.gens()
```

Now we compute a generic formal power series solution to this equation:

```python
sol = R_C(H[3][0]).power_series_solution(u_c[0], {i: a[i] for i in range(len(a))})
```

```python
((3/2)*sol * sol.derivative() + (1/4)*sol.derivative(times=3))[:15]
```

We can see that this is always a solution to our equation, hence that formal power series must have a non-trivial centralizer.

## Example with n = 3

```python
H = dict()
for i in range(10):
    _,H[i] = almost_commuting_wilson(3,i)

P = H[0][0].parent()
C = P.constant_ring().to_sage()
u = P.gens()[:-1]
z = P.gens()[-1]
```

From equation 2, we can see that we can compute $u_2^{(i)}(0)$ for any $i \geq 4$ knowing the 8 values $u_2(0), u_2'(0), u_2''(0), u_2'''(0), u_3(0), u_3'(0), u_3''(0), u_3'''(0)$. Our goal is to find another equation that allows to compute $u_3^{(j)}(0)$ for $j \geq 4$ using the same 8 coefficients (or less).

```python
[[H[4][i].coefficient_full(ui[H[4][i].order(ui)]) for ui in u] for i in range(2)]
```

```python
(H[4][1].derivative()*2/3 - H[4][0]).orders()[:2]
```

This equation is preciesely what we were looking for. Here we can obtain higher derivtives of $u_3$ using the same 8 initial conditions as before.

```python
## equs[0] allows to compute u_2 and equs[1] allows to compute u_3
equs = [H[4][1], (H[4][1].derivative()*2/3 - H[4][0])]
```

```python
R_C = H[4][0].parent().add_constants(*[f"a_{i}_{j}" for i in range(2,4) for j in range(4)])
u_c = R_C.gens()[:-1]
C_e = R_C.constant_ring()
a = C_e.gens()
```

```python
hs = [(equ.coefficient_full(ui[equ.order(ui)]), equ.coefficient_full(ui[equ.order(ui)])*ui[equ.order(ui)] - equ) for (equ,ui) in zip(equs, u_c)]

lhs = [C_e(hs[i][0]) for i in range(len(hs))]
rhs = [hs[i][1] for i in range(len(hs))]
```

```python
base = {(i,j) : C_e(f"a_{i}_{j}") for i in range(2,4) for j in range(4)}
coeffs = {f"u{i}_{j}": C_e(base[(i,j)]) for i in range(2,4) for j in range(4)}
```

```python
reg = r"u(\d+)_(\d+)"
def get_coeff(n,m):
    if n < 2 or n > 4:
        raise ValueError("Incorrect value for index 'n'")
    elif m < 0:
        return C_e.zero()

    current_var = f"u{n}_{m}"

    if not current_var in coeffs:
        RHS = rhs[n-2].derivative(times=m-equs[n-2].order(u[n-2])) # rhs of the equation u_n^{(m)} = ...
        RHS_poly = RHS.to_sage().polynomial()
        var_names = [str(v) for v in RHS_poly.variables()]
        required = [match(reg, var).groups() for var in var_names]
        # we compute previous required elements
        for (i,j) in required:
            get_coeff(int(i),int(j))
        RHS_eval = C_e(RHS_poly(**{k:v for (k,v) in coeffs.items() if k in var_names}))
        coeffs[current_var] = RHS_eval/lhs[n-2]
    return coeffs[current_var]
```

```python
LS = LaurentSeries(C_e, 't')
sols = [LS.element_class(LS, coefficient_map = lambda n : get_coeff(i,n)/factorial(n) if n >= 0 else get_coeff(i,n), order=0) for i in range(2,4)]
```

```python
sols = [
    LS.element_class(LS, coefficient_map = lambda n : get_coeff(2,n)/factorial(n) if n >= 0 else get_coeff(2,n), order=0),
    LS.element_class(LS, coefficient_map = lambda n : get_coeff(3,n)/factorial(n) if n >= 0 else get_coeff(3,n), order=0)
]
```

```python
import re
reg_f = r"u_(\d+)_(\d+)"
reg_f = re.compile(reg_f)
def replacer(match):
    n,m = [int(el) for el in match.groups()]
    if m == 0:
        return f"sols[{n-2}]"
    elif m == 1:
        return f"sols[{n-2}].derivative()"
    else:
        return f"sols[{n-2}].derivative(times={m})"
```

```python
def evaluate_equation(equation, regex, replacer):
    locals_dict = {"sols" : sols}
    command = regex.sub(replacer, str(equation))
    res = sage_eval(command, locals=locals_dict)
    return res
```
