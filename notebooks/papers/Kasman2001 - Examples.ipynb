### Example subsection 2.3.2

They show that the operator with constant coefficients $p(\partial_x,\partial_y) = (\partial_x\partial_y - \lambda)^3$ can be written as a product $L\circ K$ where
$$K = xy (\partial_x\partial_y - \lambda) \cdot \frac{1}{xy}$$
and
$$L = \partial_x^2\partial_y^2 + \frac{1}{x}\partial_x\partial_y^2 - \frac{1}{x^2}\partial_y^2 + \frac{1}{y}\partial_x^2\partial_y + \frac{1-2\lambda xy}{xy}\partial_x\partial_y - \frac{1+\lambda xy}{x^2y}\partial_y-\frac{1}{y^2}\partial_x^2 - \frac{1+\lambda xy}{xy^2}\partial_y + \lambda^2 + \frac{1}{x^2y^2}+ \frac{\lambda}{xy}$$

```python
p = (u[(1,1)]-l*u[0]).sym_power(3, u)
```

```python
K = (x*y*u[(1,1)] - l*x*y*u[0]).dot((u[0])/(x*y), u)
K
```

```python
L = u[(2,2)] + (1/x)*u[(1,2)] - (1/x^2)*u[(0,2)] + 1/y*u[(2,1)] + (1-2*l*x*y)/(x*y)*u[(1,1)] - (1+l*x*y)/(x^2*y)*u[(0,1)] - 1/y^2*u[(2,0)] - (1+l*x*y)/(x*y^2)*u[(1,0)] + (l^2 + 1/(x^2*y^2) + l/(x*y))*u[0]
L
```

```python
L.dot(K,u) == p
```

### The zero possibility (Subsection 3.3)

#### Set up

They take three PDOs that commute, and show the resultant that comes up is zero. Thes operators are:
$$L_1 = \partial_x^2 - \partial_y^2 - 1,\quad L_2 = \partial_x \circ L_1,\quad L_3 = \partial_y \circ L_1$$

```python
L1 = u[(2,0)] - u[(0,2)] - u[0]
L2 = u[(1,0)].dot(L1,u)
L3 = u[(0,1)].dot(L1,u)
for el in [L1,L2,L3]:
    show(el)
```

```python
L1.lie_bracket(L2, u), L1.lie_bracket(L3, u), L2.lie_bracket(L3, u)
```

#### The spectral operators

```python
L = spectral_operators(L1,L2,L3,names=["mu_1","mu_2","mu_3"])
CE = C.fraction_field().add_constants("mu_1","mu_2","mu_3")
```

```python
uu = L[0].parent().gen("u")
mu1, mu2, mu3 = L[0].parent().base().gens()[3:]
```

#### Some resultant relations

```python
R1 = L[0].derivative(0) - L[1]
R1
```

```python
R2 = L[0].derivative(1) - L[2]
R2
```

```python
R3 = R1.derivative(1) - R2.derivative(0)
R3
```

```python
(mu1*R3 - mu3*R1 + mu2*R2)
```

Hence, we have discovered that $(L_1 - \mu_1)\circ (L_2 - \mu_2) = (\partial_x^3 - \partial_x\partial_y^2 - \partial_x - \mu_2) \circ (L_1 - \mu_1)$, which was trivial since $L_2 = \partial_x \circ L_1$.

Similarly, we have that $L_1 \circ L_3 = (\partial_x^2\partial_y - \partial_y^3 - \partial_y - \mu_3)\circ L_1$, which is also easy to see since $L_3 = \partial_y \circ L_1$

#### Looking for a non-trivial polynomial for the operators

Let us look for what type of monomials can be the leading monomials in the polynomial relations:

###### Order 6

```python
HCP.T(6)
```

```python
for el in HCP.H(6):
    show(el)
```

Here he have possible cancellations. In fact, the solution is essentially generated by one possibility:

```python
sol,ker = HCP.cancelling_solutions(vector(ZZ,HCP.dim(6)), 6)
sol,ker
```

```python
H6 = HCP.H(6)
H6[0] - H6[1] + H6[3]
```

Let us check what happens when applying to the whole operators:

```python
candidate = HCP.from_vector(ker[0], 6)
print(candidate.order(uu))
candidate
```

We got a non-trivial operator of order 4. We need to cancel this out with the possible constructions of order 4:

```python
show(h(candidate))
sol,ker = HCP.cancelling_solutions(HCP.to_vector(h(candidate)), 4)
sol,ker
```

Here we see the following: this highest term monomial can be cancelled uniquely. We use this information to keep reducing:

```python
h(candidate - HCP.from_vector(sol, 4) - 2*mu2*L[1] + 2*mu3*L[2] + (3*mu1^2 + 2*mu1)*L[0]) 
```

```python
poly = CE(h(candidate - HCP.from_vector(sol, 4) - 2*mu2*L[1] + 2*mu3*L[2] + (3*mu1^2 + 2*mu1)*L[0]).coefficients()[0])
```

**CONCLUSION:** We found a non-trivial relation between the three operators that was not present in the Paper. We found it by *searching* possibilities and some simple linear algebra. 

We can also guarantee this is the polynomial relation with somehow a minimal property.

```python
poly
```

```python
L1.sym_power(3,u) + L1.sym_power(2, u) - L2.sym_power(2,u) + L3.sym_power(2,u)
```

```python
I = ideal(poly.numerator().wrapped)
I.change_ring(I.ring().remove_var(I.ring()('lambda_'))).dimension()
```

This polynomial relation defines a surface, what it would be the spectral variety for the three commuting operators.

### Possitive results (Subsection 3.4)

Now we have a method (or idea) that allows to compute the variety for a non-trivial case, let us check that the good example in the paper also yields a nice result with this approach.

In this example, we consider the following three operators:
$$L_1 = \partial_x^2 - \partial_y^2,\quad L_2 = y\partial_x+x\partial_y,\quad L_3 = L_1\circ L_2 - \lambda L_1.$$

#### Set up

```python
L1 = u[(2,0)] - u[(0,2)]
L2 = x*u[(0,1)] + y*u[(1,0)]
L3 = L1.dot(L2, u) - l*L1
for el in (L1,L2,L3):
    show(el)
```

```python
L1.lie_bracket(L2, u), L1.lie_bracket(L3, u), L2.lie_bracket(L3, u)
```

#### Spectral operators

```python
L = spectral_operators(L1,L2,L3,names=["mu_1","mu_2","mu_3"])
CE = C.fraction_field().add_constants("mu_1","mu_2","mu_3")
```

```python
uu = L[0].parent().gen("u")
mu1, mu2, mu3 = L[0].parent().base().gens()[3:]
```

#### Looking for non-trivial polynomial annihilating the operators

###### Order 0 - 2

```python
HCP = CommutingPDOHighest(L, uu, CE.to_sage())
```

```python
HCP.T(0)
```

```python
HCP.T(1)
```

```python
HCP.T(2)
```

###### Order 3

```python
HCP.T(3)
```

```python
sol,ker = HCP.cancelling_solutions(vector(ZZ,HCP.dim(3)), 3)
sol,ker
```

```python
H3 = HCP.H(3)
H3[0] - H3[2]
```

Let us check what happens when applying to the whole operators:

```python
candidate = HCP.from_vector(ker[0], 3)
print(candidate.order(uu))
candidate
```

```python
h(candidate - (l - mu2)*L[0] + mu1*L[1])
```

```python
poly = CE((candidate - (l - mu2)*L[0] + mu1*L[1]).coefficients()[0])
poly
```

```python
l*L1 - L1.dot(L2,u) + L3
```

```python
I = ideal(poly.numerator().wrapped)
I.change_ring(I.ring().remove_var(mu1,mu2,mu3).fraction_field()[mu1,mu2,mu3]).dimension()
```

This is precisely the polynomial obtained in the paper.
